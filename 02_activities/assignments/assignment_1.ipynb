### Classification using KNN

Let's set up our workspace and use the **Wine dataset** from `scikit-learn`. This dataset contains 178 wine samples with 13 chemical features, used to classify wines into different classes based on their origin.

The **response variable** is `class`, which indicates the type of wine. We'll use all of the chemical features to predict this response variable.

```python
# Import standard libraries
import pandas as pd
import numpy as np
import random
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import recall_score, precision_score
from sklearn.model_selection import cross_validate
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import accuracy_score
```

```python
from sklearn.datasets import load_wine 

# Load the Wine dataset
wine_data = load_wine()

# Convert to DataFrame usind pandas (renamed to pd)
wine_df = pd.DataFrame(wine_data.data, columns=wine_data.feature_names)

# Bind the 'class' (wine target) to the DataFrame
wine_df['class'] = wine_data.target

# Display the DataFrame
wine_df
```

#### **Question 2:** 
#### Standardization and data-splitting

Next, we must preform 'pre-processing' or 'data munging', to prepare our data for classification/prediction. For KNN, there are three essential steps. A first essential step is to 'standardize' the predictor variables. We can achieve this using the scaler method, provided as follows:

```python
# Select predictors (excluding the last column)
predictors = wine_df.iloc[:, :-1]

# Standardize the predictors
scaler = StandardScaler()
predictors_standardized = pd.DataFrame(scaler.fit_transform(predictors), columns=predictors.columns)

# Display the head of the standardized predictors
print(predictors_standardized.head())
predictors_standardized

outcome = wine_df['class'] # Need to add the outcome back into the dataset so we can create test and train models
wine_standardized = predictors_standardized.copy()
wine_standardized['class'] = outcome.values
```

(i) Why is it important to standardize the predictor variables?

# Answer 
When using statistical techniques like K-nearest neighbours classification, variables often have different scales (units) which can affect predictions whereby predictors with larger scales have a larger impact compared to predictors with smaller scales. To ensure each variable is comparable, it is best practice to standardize (subtracting mean from value and dividing by standard deviation) so that all predictors are on the same scale. This way the predictors can be compared on equal terms and that the model is not overly influenced by predictors with larger scales.


(ii) Why did we elect not to standard our response variable `Class`?

# Anaswer 
Class is the outcome so it does not get standardized-- this is the variable we are trying to predict so it does not enter the model as a predictor.

(iii) A second essential step is to set a random seed. Do so below (Hint: use the random.seed function). Why is setting a seed important? Is the particular seed value important? Why or why not?

# Answer 

Setting a random seed is important for splitting the dataset into a training and testing subsections in a way that is random and not biased. However, since we want the analysis to be reproducible, it is important to use the same seed (though it does not matter what seed value is selected) when reproducing the analysis.

(iv) A third essential step is to split our standardized data into separate training and testing sets. We will split into 75% training and 25% testing. The provided code randomly partitions our data, and creates linked training sets for the predictors and response variables. 

Extend the code to create a non-overlapping test set for the predictors and response variables.

```python
# set a seed for reproducibility
np.random.seed(123)

# split the data into a training and testing set. hint: use train_test_split !

wine_train, wine_test = train_test_split(wine_standardized, train_size= 0.75, shuffle=True, stratify = wine_standardized["class"])


```

#### **Question 3:**
#### Model initialization and cross-validation
We are finally set to fit the KNN model. 


Perform a grid search to tune the `n_neighbors` hyperparameter using 10-fold cross-validation. Follow these steps:

1. Initialize the KNN classifier using `KNeighborsClassifier()`.
2. Define a parameter grid for `n_neighbors` ranging from 1 to 50.
3. Implement a grid search using `GridSearchCV` with 10-fold cross-validation to find the optimal number of neighbors.
4. After fitting the model on the training data, identify and return the best value for `n_neighbors` based on the grid search results.

```python

# 1 Initialize model
knn = KNeighborsClassifier()

#Setting predictors and outcome
X_train = wine_train.drop(columns=['class'])
y_train = wine_train['class']


# 2 Define parameter grid
param_grid = {
    'n_neighbors': np.arange(1, 51)
}

# 3 Initialize and fit our grid search
sacr_gridsearch = GridSearchCV(
    estimator = knn,
    param_grid = param_grid,
    cv = 10,
)

# 4 Fit the model
sacr_gridsearch.fit(X_train, y_train)


# Best results
best_k = sacr_gridsearch.best_params_['n_neighbors']
print("Best n_neighbors:", best_k)

```

# Answer
Based on the grid search, the best number of neighbours (value of k) is 7

#### **Question 4:**
#### Model evaluation

Using the best value for `n_neighbors`, fit a KNN model on the training data and evaluate its performance on the test set using `accuracy_score`.

```python
#Setting predictors and outcome for the TEST data

knn = KNeighborsClassifier(n_neighbors=7) # based on the previous step where 7 was the best option

X_test = wine_test.drop(columns=['class'])
y_test = wine_test['class']

knn.fit(X_test,y_test)

```

```python
# VERSION 1: Splitting into 10 folds and then getting 10 different accuracy scores
returned_dictionary = cross_validate(
    estimator=knn,
    cv=10,
    X=X_test,
    y=y_test,
)

accuracy_testing = pd.DataFrame(returned_dictionary)
accuracy_testing

accuracy_testing.agg(["mean", "sem"])

```

```python
# VERSION 2: Splitting into 5 folds and then getting 10 different accuracy scores
returned_dictionary = cross_validate(
    estimator=knn,
    cv=5,
    X=X_test,
    y=y_test,
)

accuracy_testing = pd.DataFrame(returned_dictionary)
accuracy_testing

accuracy_testing.agg(["mean", "sem"])
```

# Answer

The model accuracy with k=7 based on the test data is high when using 10 and 5 fold cross validation, as it is about 0.98 or 98% in both cases (mean value) with a standard error of about 0.02-0.03. Thus, I am satisfied with the accuracy of this model.
